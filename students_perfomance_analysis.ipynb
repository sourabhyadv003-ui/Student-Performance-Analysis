```python
plt.scatter(df["study_hours_per_day"],
            df["exam_score"])
plt.xlabel("Study Hours Per Day")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

plt.show()
```

```python
plt.scatter(df["attendance_pct"], df["exam_score"])
plt.xlabel("Attendance Percentage")
plt.ylabel("Exam Score")
plt.title("Attendance vs Exam Score")

plt.show()
```

```python
city_avg = df.groupby("city")["exam_score"].mean().sort_values(ascending=False)
    
city_avg.plot(kind="bar")
                      
plt.xlabel("city")
plt.ylabel("Average Exam Score")
plt.title("City-Wise Average Exam Score")
plt.xticks(rotation=45)

plt.show()                      
```

```python
grade_counts = df["grade"].value_counts()
print(grade_counts)
grade_counts.plot(kind="bar")
plt.xlabel("Grade")
plt.ylabel("Number Of Students")
plt.title("Grade Distribution")
            
plt.show()

      
```

```python
correlation = df[['attendance_pct',
       'study_hours_per_day', 'assignment_score', 'exam_score']].corr()
sns.heatmap(correlation,annot=True,fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
study_category = pd.cut(df["study_hours_per_day"], bins=[0,3,6,10],labels=["Low","Medium","High"])
study_avg = df.groupby(study_category,observed = True)["exam_score"].mean()
ax = study_avg.plot(kind="bar")
plt.xlabel("Study Category")
plt.ylabel("Average Exam Score")
plt.title("Study Hours Category Vs Average Exam Score")

for i, value in enumerate(study_avg):
    ax.text(i, value,f"{value:.1f}",ha="center", va ="bottom")
plt.show()
```

```python
plt.hist(df["exam_score"], bins = 10)

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Exam Score Distribution")

plt.show()
```
